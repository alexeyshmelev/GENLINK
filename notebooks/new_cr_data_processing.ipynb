```python
# Self-contained dataset builder for latest GENLINK format.
#
# Output without PCA:
# node_id1,node_id2,label_id1,label_id2,ibd_sum,ibd_n
#
# Output with PCA:
# node_id1,node_id2,label_id1,label_id2,ibd_sum,ibd_n,
# node1_PC1,...,node1_PC20,node2_PC1,...,node2_PC20
#
# Logic:
# - keep only nodes with Central-Russia > 80
# - keep all edges where both endpoints are Central-Russia > 80
# - label only nodes that pass the 3-ancestor rule
# - all other included CR > 80 nodes become "masked"
# - tree file is used only for labels; nodes absent from tree are still kept as masked
# - PCA file is optional
# - if PCA is missing for a node, its PCA vector is filled with zeros

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

# --------------------
# Paths
# --------------------
GRAPH_PATH = "/disk/10tb/home/shmelev/New_CR_2025/df_anonymized.txt"
PRED_PATH  = "/disk/10tb/home/shmelev/New_CR_2025/samples_2347658_pivot_ancestry_anonymized.csv"
TREE_PATH  = "/disk/10tb/home/shmelev/New_CR_2025/trees_no_dups.csv"

# Optional PCA file.
# Set to None if you do not want PCA columns.
PCA_PATH = "/disk/10tb/home/shmelev/New_CR_2025/anonymized_id_PC1-PC20_chiptype.tsv"
# Example:
# PCA_PATH = "/disk/10tb/home/shmelev/New_CR_2025/pca_coordinates.csv"

OUT_PATH = "/disk/10tb/home/shmelev/New_CR_2025/CR_gt80_all_unlabeled_PCA.csv"

# --------------------
# Settings
# --------------------
CR_THRESHOLD = 80.0

TARGET_GROUPS = [
    "Northen Russians",
    "Southern Russians",
    "Belarusians",
    "Ukranians",
]

MASK_LABEL = "masked"

PCA_ID_COL = "anonymized_id"
PCA_DIM = 20
PCA_COLS = [f"PC{i}" for i in range(1, PCA_DIM + 1)]


def normalize_string_id(series):
    return series.astype("string").str.strip()


def print_stats_table(title, df):
    print(f"\n{title}")
    if df.empty:
        print("Empty")
    else:
        print(df.to_string(index=False))


# --------------------
# Stage A: Load
# --------------------
print("\n[Stage A] Loading files...")

pd_graph = pd.read_csv(
    GRAPH_PATH,
    dtype={"node_id1": "string", "node_id2": "string"},
)

predicted_ancestry = pd.read_csv(
    PRED_PATH,
    dtype={"node_tubeid": "string"},
)

ancestry_information = pd.read_csv(
    TREE_PATH,
    dtype={"node_tubeid": "string"},
)

pd_graph["node_id1"] = normalize_string_id(pd_graph["node_id1"])
pd_graph["node_id2"] = normalize_string_id(pd_graph["node_id2"])
predicted_ancestry["node_tubeid"] = normalize_string_id(predicted_ancestry["node_tubeid"])
ancestry_information["node_tubeid"] = normalize_string_id(ancestry_information["node_tubeid"])

required_graph_cols = {"node_id1", "node_id2", "ibd_sum", "ibd_n"}
required_pred_cols = {"node_tubeid", "Central-Russia"}
required_tree_cols = {"node_tubeid", "group"}

missing = (
    (required_graph_cols - set(pd_graph.columns))
    | (required_pred_cols - set(predicted_ancestry.columns))
    | (required_tree_cols - set(ancestry_information.columns))
)

if missing:
    raise ValueError(f"Missing required columns: {sorted(missing)}")

pd_graph["ibd_sum"] = pd.to_numeric(pd_graph["ibd_sum"], errors="raise")
pd_graph["ibd_n"] = pd.to_numeric(pd_graph["ibd_n"], errors="raise")

predicted_ancestry["Central-Russia"] = pd.to_numeric(
    predicted_ancestry["Central-Russia"],
    errors="coerce",
)

print(
    "pd_graph:", pd_graph.shape,
    "| predicted_ancestry:", predicted_ancestry.shape,
    "| ancestry_information:", ancestry_information.shape,
)

# --------------------
# Stage 0: Keep only graph nodes that have prediction info
# --------------------
print("\n[Stage 0] Intersect graph nodes with prediction file...")

graph_nodes = pd.Index(
    pd.concat([pd_graph["node_id1"], pd_graph["node_id2"]], ignore_index=True)
    .dropna()
    .unique(),
    dtype="string",
)

pred_nodes = pd.Index(
    predicted_ancestry["node_tubeid"].dropna().unique(),
    dtype="string",
)

graph_pred_nodes = graph_nodes.intersection(pred_nodes)

print("graph_pred_nodes:", len(graph_pred_nodes))

pd_graph_common = pd_graph[
    pd_graph["node_id1"].isin(graph_pred_nodes)
    & pd_graph["node_id2"].isin(graph_pred_nodes)
].copy()

pred_common = predicted_ancestry[
    predicted_ancestry["node_tubeid"].isin(graph_pred_nodes)
].copy()

print(
    "pd_graph_common:", pd_graph_common.shape,
    "| pred_common:", pred_common.shape,
)

# --------------------
# Stage 1: Central-Russia > 80
# --------------------
print(f"\n[Stage 1] Filter Central-Russia > {CR_THRESHOLD}...")

cr_nodes = pd.Index(
    pred_common.loc[
        pred_common["Central-Russia"] > CR_THRESHOLD,
        "node_tubeid",
    ].dropna().unique(),
    dtype="string",
)

print("cr_nodes:", len(cr_nodes))

# Keep all CR-CR edges.
# This includes labeled-labeled, labeled-masked, and masked-masked edges.
pd_graph_cr = pd_graph_common[
    pd_graph_common["node_id1"].isin(cr_nodes)
    & pd_graph_common["node_id2"].isin(cr_nodes)
].copy()

# Tree is used only for assigning labels.
# Nodes without tree rows will still be kept as masked if they appear in pd_graph_cr.
tree_cr = ancestry_information[
    ancestry_information["node_tubeid"].isin(cr_nodes)
].copy()

print("pd_graph_cr:", pd_graph_cr.shape, "| tree_cr:", tree_cr.shape)

# --------------------
# Stage 2: Select labeled nodes by 3-ancestor rule
# --------------------
print("\n[Stage 2] Select labeled nodes by 3-ancestor rule...")

main_node_label = {}

for node, sub in tqdm(
    tree_cr.groupby("node_tubeid", sort=False),
    desc="Selecting labeled nodes",
    unit="node",
):
    groups = sub["group"].dropna().to_numpy()

    # We expect exactly 4 ancestor rows.
    # If not, the node remains masked.
    if groups.shape[0] != 4:
        continue

    uniq, cnts = np.unique(groups, return_counts=True)

    maj_idx = int(np.argmax(cnts))
    maj_group = str(uniq[maj_idx])
    maj_cnt = int(cnts[maj_idx])

    # 3-ancestor rule:
    # at least 3 of 4 ancestors must belong to the same target group.
    if maj_cnt >= 3 and maj_group in TARGET_GROUPS:
        main_node_label[node] = maj_group

main_nodes = pd.Index(list(main_node_label.keys()), dtype="string")

print("Labeled nodes selected before edge filtering:", len(main_nodes))

# --------------------
# Stage 3: Final graph = all CR-CR edges
# --------------------
print("\n[Stage 3] Build final CR > 80 graph with all masked nodes...")

pd_graph_final = pd_graph_cr.copy()

included_nodes = pd.Index(
    pd.concat(
        [pd_graph_final["node_id1"], pd_graph_final["node_id2"]],
        ignore_index=True,
    ).dropna().unique(),
    dtype="string",
)

included_labeled_nodes = included_nodes.intersection(main_nodes)
included_masked_nodes = included_nodes.difference(main_nodes)

cr_nodes_without_cr_edges = cr_nodes.difference(included_nodes)

print("Final CR-CR edges:", pd_graph_final.shape[0])
print("Unique included nodes:", len(included_nodes))
print("Included labeled nodes:", len(included_labeled_nodes))
print("Included masked nodes:", len(included_masked_nodes))
print("CR > 80 nodes not written because they have no CR-CR edge:", len(cr_nodes_without_cr_edges))

# --------------------
# Stage 4: Add GENLINK endpoint labels
# --------------------
print("\n[Stage 4] Add label_id1 and label_id2...")

node_label = {node: MASK_LABEL for node in included_nodes}

for node, label in main_node_label.items():
    if node in node_label:
        node_label[node] = label

pd_graph_final["label_id1"] = pd_graph_final["node_id1"].map(node_label)
pd_graph_final["label_id2"] = pd_graph_final["node_id2"].map(node_label)

if pd_graph_final["label_id1"].isna().any() or pd_graph_final["label_id2"].isna().any():
    bad = pd_graph_final[
        pd_graph_final["label_id1"].isna()
        | pd_graph_final["label_id2"].isna()
    ][["node_id1", "node_id2", "label_id1", "label_id2"]].head(10)

    raise RuntimeError(f"Some endpoint labels were not assigned:\n{bad}")

pd_graph_final = pd_graph_final[
    ["node_id1", "node_id2", "label_id1", "label_id2", "ibd_sum", "ibd_n"]
].copy()

# Node-level summary used for final reporting.
node_summary = pd.DataFrame({"node": included_nodes.astype("string")})
node_summary["label"] = node_summary["node"].map(node_label)
node_summary["node_type"] = np.where(
    node_summary["label"] == MASK_LABEL,
    "masked",
    "labeled",
)

# --------------------
# Stage 5: Optional PCA coordinates
# --------------------
if PCA_PATH is not None:
    print("\n[Stage 5] Add PCA coordinates...")

    pca_df = pd.read_csv(
        PCA_PATH,
        dtype={PCA_ID_COL: "string"},
        sep=None,
        engine="python",
    )

    if PCA_ID_COL not in pca_df.columns:
        raise ValueError(f"PCA file must contain ID column: {PCA_ID_COL}")

    missing_pca_cols = [c for c in PCA_COLS if c not in pca_df.columns]
    if missing_pca_cols:
        raise ValueError(f"PCA file is missing columns: {missing_pca_cols}")

    pca_df[PCA_ID_COL] = normalize_string_id(pca_df[PCA_ID_COL])

    pca_df = pca_df[[PCA_ID_COL, *PCA_COLS]].copy()
    pca_df = pca_df.drop_duplicates(subset=PCA_ID_COL, keep="first")

    for c in PCA_COLS:
        pca_df[c] = pd.to_numeric(pca_df[c], errors="raise").astype(np.float32)

    pca_nodes = pd.Index(
        pca_df[PCA_ID_COL].dropna().unique(),
        dtype="string",
    )

    node_summary["has_pca"] = node_summary["node"].isin(pca_nodes)

    missing_pca_nodes = pd.Index(
        node_summary.loc[~node_summary["has_pca"], "node"],
        dtype="string",
    )

    print("Nodes with PCA:", int(node_summary["has_pca"].sum()))
    print("Nodes without PCA, will be filled with zeros:", len(missing_pca_nodes))

    if len(missing_pca_nodes) > 0:
        print("First nodes without PCA:", missing_pca_nodes[:10].tolist())

    node1_pca_cols = [f"node1_PC{i}" for i in range(1, PCA_DIM + 1)]
    node2_pca_cols = [f"node2_PC{i}" for i in range(1, PCA_DIM + 1)]

    pca_node1 = pca_df.rename(
        columns={
            PCA_ID_COL: "node_id1",
            **{f"PC{i}": f"node1_PC{i}" for i in range(1, PCA_DIM + 1)},
        }
    )

    pca_node2 = pca_df.rename(
        columns={
            PCA_ID_COL: "node_id2",
            **{f"PC{i}": f"node2_PC{i}" for i in range(1, PCA_DIM + 1)},
        }
    )

    pd_graph_final = pd_graph_final.merge(
        pca_node1,
        on="node_id1",
        how="left",
    )

    pd_graph_final = pd_graph_final.merge(
        pca_node2,
        on="node_id2",
        how="left",
    )

    # Fill missing PCA coordinates with zeros.
    for c in node1_pca_cols + node2_pca_cols:
        pd_graph_final[c] = pd_graph_final[c].fillna(0.0).astype(np.float32)

    final_cols = [
        "node_id1",
        "node_id2",
        "label_id1",
        "label_id2",
        "ibd_sum",
        "ibd_n",
        *node1_pca_cols,
        *node2_pca_cols,
    ]

    pd_graph_final = pd_graph_final[final_cols].copy()

    print("Added PCA columns:", len(node1_pca_cols) + len(node2_pca_cols))

else:
    print("\n[Stage 5] PCA_PATH is None, saving without PCA columns...")
    node_summary["has_pca"] = False

# --------------------
# Stage 6: Save
# --------------------
print("\n[Stage 6] Saving...")

pd_graph_final.to_csv(OUT_PATH, index=False)

print("Saved:", OUT_PATH)

# --------------------
# Stage 7: Final stats
# --------------------
print("\n[Stage 7] Final dataset stats...")

total_nodes = len(node_summary)
labeled_nodes = int((node_summary["node_type"] == "labeled").sum())
masked_nodes = int((node_summary["node_type"] == "masked").sum())

nodes_with_pca = int(node_summary["has_pca"].sum())
nodes_without_pca = int((~node_summary["has_pca"]).sum())

overall_stats = pd.DataFrame(
    [
        {
            "metric": "final_edges",
            "value": int(pd_graph_final.shape[0]),
            "proportion": np.nan,
        },
        {
            "metric": "final_columns",
            "value": int(pd_graph_final.shape[1]),
            "proportion": np.nan,
        },
        {
            "metric": "total_nodes",
            "value": total_nodes,
            "proportion": 1.0,
        },
        {
            "metric": "labeled_nodes",
            "value": labeled_nodes,
            "proportion": labeled_nodes / total_nodes if total_nodes else 0.0,
        },
        {
            "metric": "masked_nodes",
            "value": masked_nodes,
            "proportion": masked_nodes / total_nodes if total_nodes else 0.0,
        },
        {
            "metric": "nodes_with_pca",
            "value": nodes_with_pca,
            "proportion": nodes_with_pca / total_nodes if total_nodes else 0.0,
        },
        {
            "metric": "nodes_without_pca",
            "value": nodes_without_pca,
            "proportion": nodes_without_pca / total_nodes if total_nodes else 0.0,
        },
        {
            "metric": "cr_nodes_without_cr_edges_not_written",
            "value": int(len(cr_nodes_without_cr_edges)),
            "proportion": len(cr_nodes_without_cr_edges) / len(cr_nodes) if len(cr_nodes) else 0.0,
        },
    ]
)

print_stats_table("Overall stats", overall_stats)

stats_by_node_type = (
    node_summary
    .groupby("node_type", dropna=False)
    .agg(
        total_nodes=("node", "size"),
        nodes_with_pca=("has_pca", "sum"),
    )
    .reset_index()
)

stats_by_node_type["nodes_without_pca"] = (
    stats_by_node_type["total_nodes"] - stats_by_node_type["nodes_with_pca"]
)

stats_by_node_type["category_proportion_of_all_nodes"] = (
    stats_by_node_type["total_nodes"] / total_nodes if total_nodes else 0.0
)

stats_by_node_type["with_pca_proportion_inside_category"] = np.where(
    stats_by_node_type["total_nodes"] > 0,
    stats_by_node_type["nodes_with_pca"] / stats_by_node_type["total_nodes"],
    0.0,
)

stats_by_node_type["without_pca_proportion_inside_category"] = np.where(
    stats_by_node_type["total_nodes"] > 0,
    stats_by_node_type["nodes_without_pca"] / stats_by_node_type["total_nodes"],
    0.0,
)

print_stats_table("PCA stats by node type", stats_by_node_type)

label_order = TARGET_GROUPS + [MASK_LABEL]

stats_by_label = (
    node_summary
    .groupby("label", dropna=False)
    .agg(
        total_nodes=("node", "size"),
        nodes_with_pca=("has_pca", "sum"),
    )
)

stats_by_label = stats_by_label.reindex(label_order, fill_value=0).reset_index()

stats_by_label["nodes_without_pca"] = (
    stats_by_label["total_nodes"] - stats_by_label["nodes_with_pca"]
)

stats_by_label["category_proportion_of_all_nodes"] = np.where(
    total_nodes > 0,
    stats_by_label["total_nodes"] / total_nodes,
    0.0,
)

stats_by_label["with_pca_proportion_inside_category"] = np.where(
    stats_by_label["total_nodes"] > 0,
    stats_by_label["nodes_with_pca"] / stats_by_label["total_nodes"],
    0.0,
)

stats_by_label["without_pca_proportion_inside_category"] = np.where(
    stats_by_label["total_nodes"] > 0,
    stats_by_label["nodes_without_pca"] / stats_by_label["total_nodes"],
    0.0,
)

print_stats_table("PCA stats by label category", stats_by_label)

pd_graph_final.head()
```

```python
# Self-contained cell (rewritten): same pipeline, but DO NOT include neighbor–neighbor edges.
# Keep only edges that touch at least one main node.
# Adds stage printing + tqdm progress bars.

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

# --------------------
# Paths
# --------------------
GRAPH_PATH = "/disk/10tb/home/shmelev/New_CR_2025/df_anonymized.txt"
PRED_PATH  = "/disk/10tb/home/shmelev/New_CR_2025/samples_2347658_pivot_ancestry_anonymized.csv"
TREE_PATH  = "/disk/10tb/home/shmelev/New_CR_2025/trees_no_dups.csv"

OUT_PATH   = "/disk/10tb/home/shmelev/New_CR_2025/CR_gt80_semi_unlabeled_all_masks.csv"

# --------------------
# Load
# --------------------
print("\n[Stage A] Loading files...")
pd_graph = pd.read_csv(GRAPH_PATH, dtype={"node_id1": "string", "node_id2": "string"})
predicted_ancestry = pd.read_csv(PRED_PATH, dtype={"node_tubeid": "string"})
ancestry_information = pd.read_csv(TREE_PATH, dtype={"node_tubeid": "string"})

# Required columns check
required_graph_cols = {"node_id1", "node_id2", "ibd_sum", "ibd_n"}
required_pred_cols  = {"node_tubeid", "Central-Russia"}
required_tree_cols  = {"node_tubeid", "group"}
missing = (required_graph_cols - set(pd_graph.columns)) | (required_pred_cols - set(predicted_ancestry.columns)) | (required_tree_cols - set(ancestry_information.columns))
if missing:
    raise ValueError(f"Missing required columns: {sorted(missing)}")

print("pd_graph:", pd_graph.shape, "| predicted_ancestry:", predicted_ancestry.shape, "| ancestry_information:", ancestry_information.shape)

# --------------------
# 0) Keep only nodes common to ALL three sources
# --------------------
print("\n[Stage 0] Intersect nodes across graph/pred/tree...")
graph_nodes = pd.Index(
    pd.concat([pd_graph["node_id1"], pd_graph["node_id2"]], ignore_index=True).dropna().unique(),
    dtype="string",
)
pred_nodes  = pd.Index(predicted_ancestry["node_tubeid"].dropna().unique(), dtype="string")
tree_nodes  = pd.Index(ancestry_information["node_tubeid"].dropna().unique(), dtype="string")
common_nodes = graph_nodes.intersection(pred_nodes).intersection(tree_nodes)
print("common_nodes:", len(common_nodes))

pd_graph_common = pd_graph[pd_graph["node_id1"].isin(common_nodes) & pd_graph["node_id2"].isin(common_nodes)].copy()
pred_common = predicted_ancestry[predicted_ancestry["node_tubeid"].isin(common_nodes)].copy()
tree_common = ancestry_information[ancestry_information["node_tubeid"].isin(common_nodes)].copy()
print("pd_graph_common:", pd_graph_common.shape, "| pred_common:", pred_common.shape, "| tree_common:", tree_common.shape)

# --------------------
# 1) Central-Russia prediction > 80%
# --------------------
print("\n[Stage 1] Filter Central-Russia > 80...")
cr_nodes = pd.Index(pred_common.loc[pred_common["Central-Russia"] > 80.0, "node_tubeid"].unique(), dtype="string")
print("cr_nodes:", len(cr_nodes))

pd_graph_cr = pd_graph_common[pd_graph_common["node_id1"].isin(cr_nodes) & pd_graph_common["node_id2"].isin(cr_nodes)].copy()
tree_cr = tree_common[tree_common["node_tubeid"].isin(cr_nodes)].copy()
print("pd_graph_cr:", pd_graph_cr.shape, "| tree_cr:", tree_cr.shape)

# --------------------
# 2) Select main nodes (simple loop version)
# --------------------
print("\n[Stage 2] Select main nodes...")
unique_nodes = tree_cr["node_tubeid"].dropna().unique()

main_nodes_list = []
for node in tqdm(unique_nodes, desc="Selecting main nodes", unit="node"):
    sub = tree_cr[tree_cr["node_tubeid"] == node]
    groups = sub["group"].dropna().to_numpy()

    # Must have exactly 4 ancestor rows
    if groups.shape[0] != 4:
        continue

    uniq, cnts = np.unique(groups, return_counts=True)

    # allow <=2 distinct groups total
    if uniq.shape[0] > 2:
        continue

    # majority group / count
    maj_idx = int(np.argmax(cnts))
    maj_group = uniq[maj_idx]
    maj_cnt = int(cnts[maj_idx])

    # main-node condition
    if (maj_cnt >= 3) and (maj_group != "Unknown"):
        main_nodes_list.append(node)

main_nodes = pd.Index(main_nodes_list, dtype="string")
print("Main nodes:", len(main_nodes))

# --------------------
# 3) Add only nodes that connect to main nodes by at least one edge
#    IMPORTANT: keep ONLY edges touching main nodes (NO induced neighbor–neighbor edges)
# --------------------
print("\n[Stage 3] Keep only edges that touch a main node (no neighbor-neighbor edges)...")
edges_touching_main = pd_graph_cr[
    pd_graph_cr["node_id1"].isin(main_nodes) | pd_graph_cr["node_id2"].isin(main_nodes)
].copy()

included_nodes = pd.Index(
    pd.concat([edges_touching_main["node_id1"], edges_touching_main["node_id2"]], ignore_index=True).dropna().unique(),
    dtype="string",
)

# FINAL EDGE SET: only edges touching main (this prevents neighbor–neighbor edges)
pd_graph_final = edges_touching_main.copy()

print("edges_touching_main:", edges_touching_main.shape)
print("pd_graph_final:", pd_graph_final.shape)
print("Unique included nodes:", len(included_nodes))

# Safety check: every edge must touch a main node
touches_main_mask = pd_graph_final["node_id1"].isin(main_nodes) | pd_graph_final["node_id2"].isin(main_nodes)
if not bool(touches_main_mask.all()):
    bad = pd_graph_final.loc[~touches_main_mask, ["node_id1", "node_id2"]].head(10)
    raise RuntimeError(f"Found edges not touching main nodes (showing up to 10):\n{bad}")

# --------------------
# 4) Targets: counts of ancestors in each group for each node
#    Add them for BOTH endpoints, but name them label1_{group}, label2_{group}.
#    label1_* corresponds to node_id1, label2_* corresponds to node_id2.
# --------------------
print("\n[Stage 4] Build target labels for both endpoints...")
TARGET_GROUPS = ["Northen Russians", "Southern Russians", "Belarusians", "Ukranians"]

# counts_dict[node][group] = count (only for included_nodes)
counts_dict = {node: {g: 0 for g in TARGET_GROUPS} for node in included_nodes}

tree_nodes_arr = tree_cr["node_tubeid"].to_numpy()
tree_groups_arr = tree_cr["group"].to_numpy()

for node, grp in tqdm(zip(tree_nodes_arr, tree_groups_arr), total=len(tree_cr), desc="Counting ancestor groups", unit="row"):
    if (node in counts_dict) and (grp in TARGET_GROUPS):
        counts_dict[node][grp] += 1

# Prepare output columns (requested naming)
LABEL1_COLS = [f"label1_{g}" for g in TARGET_GROUPS]
LABEL2_COLS = [f"label2_{g}" for g in TARGET_GROUPS]

for c in LABEL1_COLS + LABEL2_COLS:
    pd_graph_final[c] = 0

# Fill columns with explicit loops
node1_arr = pd_graph_final["node_id1"].to_numpy()
node2_arr = pd_graph_final["node_id2"].to_numpy()
n = len(pd_graph_final)

label1_arrays = {c: np.zeros(n, dtype=np.int32) for c in LABEL1_COLS}
label2_arrays = {c: np.zeros(n, dtype=np.int32) for c in LABEL2_COLS}

for i in tqdm(range(n), desc="Writing edge labels", unit="edge"):
    u = node1_arr[i]
    v = node2_arr[i]

    du = counts_dict[u]
    dv = counts_dict[v]

    # node_id1 -> label1_*
    for g, c in zip(TARGET_GROUPS, LABEL1_COLS):
        label1_arrays[c][i] = du[g]

    # node_id2 -> label2_*
    for g, c in zip(TARGET_GROUPS, LABEL2_COLS):
        label2_arrays[c][i] = dv[g]

for c in LABEL1_COLS:
    pd_graph_final[c] = label1_arrays[c]
for c in LABEL2_COLS:
    pd_graph_final[c] = label2_arrays[c]

# Reorder columns:
# node_id1, node_id2, label1_targets..., label2_targets..., ibd_sum, ibd_n
pd_graph_final = pd_graph_final[["node_id1", "node_id2", *LABEL1_COLS, *LABEL2_COLS, "ibd_sum", "ibd_n"]].copy()

# --------------------
# Save
# --------------------
print("\n[Stage 5] Saving...")
pd_graph_final.to_csv(OUT_PATH, index=False)

print("Saved:", OUT_PATH)
print("Final edges:", pd_graph_final.shape[0])
print("Unique included nodes:", len(included_nodes))
print("Main nodes:", len(main_nodes))

pd_graph_final.head()
```
