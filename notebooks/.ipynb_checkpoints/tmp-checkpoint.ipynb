```python
from pathlib import Path
import json

import matplotlib.pyplot as plt
import numpy as np


ROOT = Path("/home/jovyan/GENLINK/downstream_tasks/runs/real_data_no_mask")
FIGURE_DIR = Path("/home/jovyan/GENLINK/pictures/PICTURES_ES")
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

DATASETS = [
    "FINAL_2nd_degree_3_plus_1_labeled_min10_without_pca",
    "FINAL_2nd_degree_3_plus_1_labeled_min8_without_pca",
    "FINAL_2nd_degree_4_of_4_labeled_min8_without_pca",
    "FINAL_all_degree_same_labeled_min8_without_pca",
    "FINAL_2nd_degree_3_plus_1_labeled_min8_with_pca",
]

CLASSES = [
    "Northen Russians",
    "Southern Russians",
    "Ukranians",
    "Belarusians",
]


for dataset in DATASETS:
    dataset_path = ROOT / dataset

    # model -> class -> scores across splits
    model_scores = {}

    for results_file in dataset_path.glob("*/results.json"):
        run_name = results_file.parent.name
        model_name = run_name.rsplit("_split_", 1)[0]

        try:
            with open(results_file) as f:
                data = json.load(f)
        except (json.JSONDecodeError, OSError):
            continue

        class_scores = data.get("class_scores")
        if not class_scores:
            continue

        for class_name in CLASSES:
            if class_name in class_scores:
                model_scores.setdefault(
                    model_name, {}
                ).setdefault(
                    class_name, []
                ).append(
                    class_scores[class_name]
                )

    if not model_scores:
        continue

    models = sorted(model_scores)
    x = np.arange(len(CLASSES))

    bar_width = 0.8 / len(models)

    fig, ax = plt.subplots(figsize=(12, 6))

    for model_idx, model in enumerate(models):
        means = []
        stds = []

        for class_name in CLASSES:
            scores = model_scores[model].get(class_name, [])

            if scores:
                means.append(np.mean(scores))
                stds.append(np.std(scores))
            else:
                means.append(np.nan)
                stds.append(0.0)

        positions = (
            x
            - 0.4
            + bar_width / 2
            + model_idx * bar_width
        )

        bars = ax.bar(
            positions,
            means,
            width=bar_width,
            yerr=stds,
            capsize=3,
            label=model,
        )

        for bar, mean, std in zip(bars, means, stds):
            if np.isnan(mean):
                continue

            ax.text(
                bar.get_x() + bar.get_width() / 2,
                bar.get_height() + std + 0.015,
                f"{mean:.3f}\n±{std:.3f}",
                ha="center",
                va="bottom",
                fontsize=8,
            )

    ax.set_title(dataset)
    ax.set_ylabel("F1 score")
    ax.set_xticks(x)
    ax.set_xticklabels(CLASSES)
    ax.set_ylim(0, 1.1)
    ax.grid(axis="y", alpha=0.3)
    ax.legend()

    fig.tight_layout()

    output_path = FIGURE_DIR / f"{dataset}_class_f1.png"
    fig.savefig(output_path, dpi=200, bbox_inches="tight")
    plt.close(fig)

    print(f"Saved: {output_path}")
```

```python
from pathlib import Path
import json

import matplotlib.pyplot as plt
import numpy as np


ROOT = Path("/home/jovyan/GENLINK/downstream_tasks/runs/real_data_real_masks")
FIGURE_DIR = Path("/home/jovyan/GENLINK/pictures/PICTURES_ES_masked")
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

# Add/remove datasets here
DATASETS = [
    "FINAL_2_all_degree_same_labeled_min8_without_pca",
    "FINAL_2_2nd_degree_3_plus_1_all_unlabeled_25pct_min8_without_pca",
    "FINAL_2_2nd_degree_3_plus_1_all_unlabeled_50pct_min8_without_pca",
    "FINAL_2_2nd_degree_3_plus_1_all_unlabeled_5pct_min8_without_pca",
    "FINAL_2_2nd_degree_3_plus_1_all_unlabeled_min8_without_pca",
    "FINAL_2_2nd_degree_3_plus_1_all_unlabeled_1pct_min8_without_pca",
    "FINAL_2_2nd_degree_3_plus_1_connected_unlabeled_min8_without_pca"
]

CLASSES = [
    "Northen Russians",
    "Southern Russians",
    "Ukranians",
    "Belarusians",
]

rows = []

for dataset in DATASETS:
    dataset_path = ROOT / dataset

    # model -> macro scores across splits
    model_macro_scores = {}

    # model -> class -> scores across splits
    model_class_scores = {}

    # Missing results.json files are automatically ignored
    for results_file in dataset_path.glob("*/results.json"):
        run_name = results_file.parent.name
        model_name = run_name.rsplit("_split_", 1)[0]

        try:
            with open(results_file) as f:
                data = json.load(f)
        except (json.JSONDecodeError, OSError):
            continue

        if "f1_macro" in data:
            model_macro_scores.setdefault(model_name, []).append(
                data["f1_macro"]
            )

        class_scores = data.get("class_scores")

        if class_scores:
            for class_name in CLASSES:
                if class_name in class_scores:
                    model_class_scores.setdefault(
                        model_name, {}
                    ).setdefault(
                        class_name, []
                    ).append(
                        class_scores[class_name]
                    )

    # Preserve the original averaged macro-F1 console output
    for model, scores in model_macro_scores.items():
        rows.append(
            {
                "dataset": dataset,
                "model": model,
                "f1_macro": np.mean(scores),
                "n_splits": len(scores),
            }
        )

    if not model_class_scores:
        print(f"No class_scores found for: {dataset}")
        continue

    models = sorted(model_class_scores)
    x = np.arange(len(CLASSES))

    bar_width = 0.8 / len(models)

    fig, ax = plt.subplots(figsize=(12, 6))

    for model_idx, model in enumerate(models):
        means = []
        stds = []

        for class_name in CLASSES:
            scores = model_class_scores[model].get(class_name, [])

            if scores:
                means.append(float(np.mean(scores)))
                stds.append(float(np.std(scores)))
            else:
                means.append(np.nan)
                stds.append(0.0)

        positions = (
            x
            - 0.4
            + bar_width / 2
            + model_idx * bar_width
        )

        bars = ax.bar(
            positions,
            means,
            width=bar_width,
            yerr=stds,
            capsize=3,
            label=model,
        )

        for bar, mean, std in zip(bars, means, stds):
            if np.isnan(mean):
                continue

            ax.text(
                bar.get_x() + bar.get_width() / 2,
                bar.get_height() + std + 0.015,
                f"{mean:.3f}\n±{std:.3f}",
                ha="center",
                va="bottom",
                fontsize=8,
            )

    ax.set_title(dataset)
    ax.set_ylabel("F1 score")
    ax.set_xticks(x)
    ax.set_xticklabels(CLASSES)
    ax.set_ylim(0, 1.1)
    ax.grid(axis="y", alpha=0.3)
    ax.legend()

    fig.tight_layout()

    output_path = FIGURE_DIR / f"{dataset}_class_f1.png"
    fig.savefig(
        output_path,
        dpi=200,
        bbox_inches="tight",
    )
    plt.close(fig)

    print(f"Saved: {output_path}")


# Original averaged macro-F1 console output
print()
print(
    f"{'Dataset':70s} "
    f"{'Model':65s} "
    f"{'F1 macro':>10s} "
    f"{'Splits':>8s}"
)

print("-" * 160)

for row in rows:
    print(
        f"{row['dataset']:70s} "
        f"{row['model']:65s} "
        f"{row['f1_macro']:10.4f} "
        f"{row['n_splits']:8d}"
    )
```

```python
import pandas as pd

# label columns
label1_cols = [
    "label1_Northen Russians",
    "label1_Southern Russians",
    "label1_Belarusians",
    "label1_Ukranians"
]

label2_cols = [
    "label2_Northen Russians",
    "label2_Southern Russians",
    "label2_Belarusians",
    "label2_Ukranians"
]

# extract classes for node_id1
node1_classes = (
    df[["node_id1"] + label1_cols]
    .melt(id_vars="node_id1", var_name="class", value_name="value")
    .query("value == 3")
    [["node_id1", "class"]]
    .rename(columns={"node_id1": "node_id"})
)

# extract classes for node_id2
node2_classes = (
    df[["node_id2"] + label2_cols]
    .melt(id_vars="node_id2", var_name="class", value_name="value")
    .query("value == 3")
    [["node_id2", "class"]]
    .rename(columns={"node_id2": "node_id"})
)

# combine, deduplicate, clean class names
node_class_df = (
    pd.concat([node1_classes, node2_classes], ignore_index=True)
    .drop_duplicates(subset="node_id")
)

node_class_df["class"] = (
    node_class_df["class"]
    .str.replace("label1_", "", regex=False)
    .str.replace("label2_", "", regex=False)
)

# class balance
class_counts = node_class_df["class"].value_counts()
class_counts
```
