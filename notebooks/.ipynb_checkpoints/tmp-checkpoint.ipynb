```python
from pathlib import Path
import json
import numpy as np

ROOT = Path("/home/jovyan/GENLINK/downstream_tasks/runs/real_data_no_mask")

# Add/remove datasets here
DATASETS = [
    "FINAL_2nd_degree_3_plus_1_labeled_min10_without_pca",
    "FINAL_2nd_degree_3_plus_1_labeled_min8_without_pca",
    "FINAL_2nd_degree_4_of_4_labeled_min8_without_pca",
    "FINAL_all_degree_same_labeled_min8_without_pca",
    "FINAL_2nd_degree_3_plus_1_labeled_min8_with_pca"
]

rows = []

for dataset in DATASETS:
    dataset_path = ROOT / dataset
    model_scores = {}

    # Missing results.json files are automatically ignored
    for results_file in dataset_path.glob("*/results.json"):
        run_name = results_file.parent.name
        model_name = run_name.rsplit("_split_", 1)[0]

        try:
            with open(results_file) as f:
                data = json.load(f)

            if "f1_macro" not in data:
                continue

            model_scores.setdefault(model_name, []).append(data["f1_macro"])

        except (json.JSONDecodeError, OSError):
            continue

    for model, scores in model_scores.items():
        rows.append({
            "dataset": dataset,
            "model": model,
            "f1_macro": np.mean(scores),
            "n_splits": len(scores),
        })


# Final output
print(
    f"{'Dataset':50s} "
    f"{'Model':65s} "
    f"{'F1 macro':>10s} "
    f"{'Splits':>8s}"
)

print("-" * 140)

for row in rows:
    print(
        f"{row['dataset']:50s} "
        f"{row['model']:65s} "
        f"{row['f1_macro']:10.4f} "
        f"{row['n_splits']:8d}"
    )
```

```python
import pandas as pd

# label columns
label1_cols = [
    "label1_Northen Russians",
    "label1_Southern Russians",
    "label1_Belarusians",
    "label1_Ukranians"
]

label2_cols = [
    "label2_Northen Russians",
    "label2_Southern Russians",
    "label2_Belarusians",
    "label2_Ukranians"
]

# extract classes for node_id1
node1_classes = (
    df[["node_id1"] + label1_cols]
    .melt(id_vars="node_id1", var_name="class", value_name="value")
    .query("value == 3")
    [["node_id1", "class"]]
    .rename(columns={"node_id1": "node_id"})
)

# extract classes for node_id2
node2_classes = (
    df[["node_id2"] + label2_cols]
    .melt(id_vars="node_id2", var_name="class", value_name="value")
    .query("value == 3")
    [["node_id2", "class"]]
    .rename(columns={"node_id2": "node_id"})
)

# combine, deduplicate, clean class names
node_class_df = (
    pd.concat([node1_classes, node2_classes], ignore_index=True)
    .drop_duplicates(subset="node_id")
)

node_class_df["class"] = (
    node_class_df["class"]
    .str.replace("label1_", "", regex=False)
    .str.replace("label2_", "", regex=False)
)

# class balance
class_counts = node_class_df["class"].value_counts()
class_counts
```

```python
import numpy as np

TARGET_GROUPS = ["Northen Russians", "Southern Russians", "Belarusians", "Ukranians"]
label_cols = [f"label1_{g}" for g in TARGET_GROUPS] + [f"label2_{g}" for g in TARGET_GROUPS]

row_sum = df[label_cols].sum(axis=1)

ok = (row_sum >= 4)
print("All rows satisfy sum>=4:", bool(ok.all()))
print("Min row sum:", int(row_sum.min()))
print("Number of bad rows:", int((~ok).sum()))

# show a few bad rows (if any)
df_bad = df.loc[~ok, ["node_id1", "node_id2", *label_cols]].head(20)
df_bad
```
